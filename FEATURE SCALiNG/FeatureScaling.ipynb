```python
import pandas as pd

df = pd.DataFrame({
    "Age": [18, 22, 25, 30, 35, 40],
    "Salary": [15000, 25000, 30000, 45000, 60000, 80000],
    "Experience": [0, 1, 2, 5, 8, 12]
})

df
```

```python
#Standard scaler
age = df[["Age"]]
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
age_scaled = scaler.fit_transform(age)
age_scaled
df["Scaled_Age"] = age_scaled
```

```python
salary = df[["Salary"]]
from sklearn.preprocessing import MinMaxScaler
mm = MinMaxScaler()
salary_scaled =mm.fit_transform(salary)
df["Scaled_Salary"] = salary_scaled
df
```

```python
experience = df[["Experience"]]
from sklearn.preprocessing import MinMaxScaler
mmScaler = MinMaxScaler()
experience_scaled =mmScaler.fit_transform(experience)
df["Scaled_Experience"] = experience_scaled
df
```
