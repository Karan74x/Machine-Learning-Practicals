```python
import numpy as np
import pandas as pd

df = pd.read_csv("./data.csv")
df.head()

df.isna().sum()

# remove the unwanted columns
df.drop(["id", "Unnamed: 32"], axis=1, inplace=True)
df.shape
```

```python
# Encoding this (Y)column df["diagnosis"] to numerical format
y = df["diagnosis"]
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_pred = le.fit_transform(y)
y = pd.DataFrame(y_pred, columns=["diagnosis"])
y
```

```python
x = df.iloc[:, 1:31]
x
```

```python
# 4. Split into training and testing data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42
)

print(len(X_train))
print(len(y_test))
print(len(X_train))
print(len(y_test))
y_test
```

```python
# 5. Create KNN object
from sklearn.neighbors import KNeighborsClassifier

# k's value between 1-16 for better accuracy
knn = KNeighborsClassifier(n_neighbors=11)


# 6. Train the model
knn.fit(X_train, y_train)

y_pred = knn.predict(X_test)
y_pred = pd.DataFrame(y_pred)
y_pred
```

```python

# 8. Accuracy
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

# currently k = 11 gives the best accuracy among others
print("Accuracy:", accuracy)
```
