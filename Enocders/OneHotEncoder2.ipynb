```python
# OneHotEncoder converts nominal (unordered) categories into binary (0/1) columns.
```

```python
import pandas as pd
import numpy as np
df = pd.read_csv("../encoding_practice.csv")

# City Column
city = df[["City"]]
city
```

```python
from sklearn.preprocessing import OneHotEncoder
# sparse_output=False returns a standard array
ohe = OneHotEncoder(sparse_output=False, drop="first")
encoded_city = ohe.fit_transform(city)
encoded_city
```

```python
# take out the unique names from city column
names_of_city = ohe.get_feature_names_out(["City"])
names_of_city
```

```python
city_encoded_df = pd.DataFrame(encoded_city, columns=names_of_city)
city_encoded_df
```
