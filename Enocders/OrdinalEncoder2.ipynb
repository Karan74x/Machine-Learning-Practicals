```python
#                     Ordinal Encoder
# OrdinalEncoder converts ordered categories into numbers.
# It preserves the natural order
```

```python
import pandas  as pd
import numpy as np

df = pd.read_csv("../encoding_practice.csv")

#  Satisfaction Example

# Select the column
satisfaction = df[["Satisfaction"]]
satisfaction

```

```python
from sklearn.preprocessing import OrdinalEncoder

# ordinalEncoder object             0        1        2          3
oe = OrdinalEncoder(categories=[["Poor", "Average", "Good", "Excellent"]])
satisfaction_encoded = oe.fit_transform(satisfaction)
satisfaction_encoded
```

```python
df["Encoded_satisfaction"] = satisfaction_encoded
df
```
