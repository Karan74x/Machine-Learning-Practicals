```python
import pandas as pd


df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Education": ["School", "Bachelor", "Master", "PhD", "Bachelor", "Master"],
    "City": ["Ahmedabad", "Mumbai", "Delhi", "Ahmedabad", "Delhi", "Mumbai"]
})
df
```

```python
# Label Encoding
gender = df[["Gender"]]
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
gender_encoded = encoder.fit_transform(gender)
gender_encoded
df["Gender_encoded"] = gender_encoded
df
df = df.drop(columns=["Gender"])
df
```

```python
# One hot Encoded
city = df[["City"]]
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop='first', sparse_output=False)
city_encoded = ohe.fit_transform(city)
city_names = ohe.get_feature_names_out(["City"])
city_names

city_encoded = pd.DataFrame(city_encoded, columns=city_names)
city_encoded

df = pd.concat([df, city_encoded], axis=1)
df = df.drop(columns=["City"])
df
```

```python
# Ordinal Encoder
education = df[["Education"]]
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[["School", "Bachelor", "Master", "PhD"]])
education_encoded = oe.fit_transform(education)
education_encoded
df["Education_Encoded"] = education_encoded
df = df.drop(columns=["Education"])
df
```
