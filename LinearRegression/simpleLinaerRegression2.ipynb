```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "CGPA": [6, 7, 8, 9, 10 ],
    "Package": [3, 4, 5, 6, 7]
})

x = df[["CGPA"]] # x  = input feature
y = df[["Package"]] # y = output/target

plt.scatter(x, y)
plt.xlabel("CGPA")
plt.ylabel("Package")
plt.title("CGPA vs Package")
```

```python
# import and create LinearRegression object
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

#train the model
lr.fit(x, y)

```

```python
m = lr.coef_
print("Slope m: ", m)
b = lr.intercept_
print("Offset b:",b)
```

```python
res = lr.predict([[9.5]])
print(res[0])
```
